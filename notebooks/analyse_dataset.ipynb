```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../dataset/creditcard.csv")

print("Dataset chargé avec succès !")
print("Dimensions :", df.shape)
```

```python
df.head()
```

```python
print("\nNoms des colonnes :")
print(df.columns.tolist())
```

```python
print("\nInformations générales :")
df.info()
```

```python
print("Valeurs manquantes :")
print(df.isnull().sum().sum())
print("Doublons :", df.duplicated().sum())
```

```python
df.describe()
```

```python
#Distribution des transactions (Normal / Fraud)
counts = df["Class"].value_counts().sort_index()

labels = ["Normale", "Fraude"]
colors = ["#2ECC71", "#E74C3C"]
plt.figure(figsize=(8, 5))
bars = plt.bar(labels, counts.values, color=colors, width=0.6)
plt.yscale("log")
plt.title("Répartition des transactions", fontsize=15)
plt.xlabel("Type de transaction")
plt.ylabel("Nombre de transactions (échelle logarithmique)")

for bar, value in zip(bars, counts.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value,
        f"{value:,}",
        ha="center",
        fontweight="bold"
    )

plt.show()
```

```python
#Pourcentage des transactions
percentages = df["Class"].value_counts(normalize=True) * 100

plt.figure(figsize=(7, 5))

plt.pie(
    percentages.values,
    labels=["Normale", "Fraude"],
    autopct="%1.2f%%",
    colors=["#2ECC71", "#E74C3C"],
    startangle=90,
    explode=(0, 0.1),
    shadow=True
)

plt.title("Pourcentage des transactions")
plt.show()
```

```python
#Distribution du montant des transactions
plt.figure(figsize=(9, 5))
sns.histplot(
    data=df,
    x="Amount",
    bins=50,
    color="#9B59B6",
    edgecolor="white"
)
plt.title("Distribution du montant des transactions")
plt.xlabel("Montant")
plt.ylabel("Fréquence")

plt.show()
```

```python
#Comparaison des montants selon la classe
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Class",
    y="Amount",
    hue="Class",
    palette={
        0: "#3498DB",
        1: "#E67E22"
    },
    showfliers=False
)

plt.title("Montant selon le type de transaction")
plt.xlabel("0 = Normale / 1 = Fraude")
plt.ylabel("Montant")

plt.show()
```

```python
#Corrélation entre les variables et Class
correlation = df.corr(numeric_only=True)["Class"]
correlation = correlation.drop("Class").sort_values()
colors = [
    "#E74C3C" if value < 0 else "#2ECC71"
    for value in correlation
]

plt.figure(figsize=(9, 8))

plt.barh(
    correlation.index,
    correlation.values,
    color=colors
)

plt.axvline(0, color="black")

plt.title("Corrélation des variables avec Class")
plt.xlabel("Coefficient de corrélation")
plt.ylabel("Variables")

plt.tight_layout()
plt.show()
```
