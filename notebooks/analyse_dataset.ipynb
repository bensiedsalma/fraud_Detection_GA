```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
df = pd.read_csv("../dataset/creditcard.csv")
df.head()
```

```python
print("Dimensions (lignes, colonnes) :", df.shape)
```

```python
print("\nNoms des colonnes :")
print(df.columns.tolist())
```

```python
print("\nInformations générales :")
df.info()
```

```python
print("Valeurs manquantes par colonne :")
display(df.isnull().sum())
print("Nombre de lignes dupliquées :", df.duplicated().sum())
```

```python
df.describe()
```

```python
counts = df["Class"].value_counts().sort_index()
percentages = df["Class"].value_counts(normalize=True).sort_index() * 100

print("Nombre de transactions par classe :")
display(counts.rename(index={0: "Normale (0)", 1: "Fraude (1)"}))
print("Pourcentage par classe :")
display(percentages.rename(index={0: "Normale (0)", 1: "Fraude (1)"}).round(3))
```

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(data=df, x="Class")

plt.yscale("log")
plt.title("Distribution des transactions")
plt.xlabel("Class (0 = Normale, 1 = Fraude)")
plt.ylabel("Nombre de transactions (échelle logarithmique)")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```
